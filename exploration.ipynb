Base case: a single date

example: 03-06-1995

`0 + 3 + 0 + 6 + 1 + 9 + 9 + 5 = 33`

The major arcana are 22 cards, since `33>22` we keep reducing: `3 + 3 = 6` 

In this case the 6th arcana is "The lovers".

Let's create a function that takes a date string (dd-mm-yyyy) and returns a number (1-22)

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
test_dates = [
    "03-04-1994"
    "13-12-1994"
    "31-03-1994"
]
```

```python
def date_to_arcana(date_str:str) -> int:
    """Assuming a valid date, retuns a number from 1 to 22

    Args:
        date_str (str): Hopefully, somebody's born date

    Returns:
        int: corresponding major arcana
    """

    # Parse the date and sum the digits for the first result
    date_digits_sum = sum([int(l) for l in date_str if l.isdigit()])

    if date_digits_sum <= 22:
        return date_digits_sum
    
    # If not valid, repeat
    sum_digits_sum = sum([int(l) for l in str(date_digits_sum) if l.isdigit()])
    return sum_digits_sum
```

```python
print(date_to_arcana("03-06-1995"))
```

```python
from datetime import datetime, timedelta

def date_range(date_a, date_b):
    start = datetime.strptime(date_a, "%d-%m-%Y")
    end = datetime.strptime(date_b, "%d-%m-%Y")

    return [
        (start + timedelta(days=i)).strftime("%d-%m-%Y")
        for i in range((end - start).days + 1)
    ]
```

Finally, we can show a chart with the distribution of arcanas

```python
arcanas = [date_to_arcana(d) for d in date_range("01-01-2020", "06-09-2026")]
len(arcanas)
```

```python

sns.countplot(x=arcanas, order=range(1, 23))

plt.xlabel("Number")
plt.ylabel("Count")
plt.title("Distribution")
plt.show()
```
